```python
# Libraries
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

plt.style.use('fivethirtyeight')
np.random.seed(777)

import warnings
warnings.filterwarnings("ignore", category = UserWarning)
warnings.filterwarnings("ignore", category = FutureWarning)
```

#### ``Loan Default Pre-Processed Data + Dimensionality Reduction (PCA)``

```python
# Loan Default Pre-Processed data + PCA
X_train = pd.read_pickle('https://raw.githubusercontent.com/sranillac/SRC_UniOvi/refs/heads/main/ML-IA/data/LoanDefault/X_Train_PCA_LoanDefault.pkl')
X_test = pd.read_pickle('https://raw.githubusercontent.com/sranillac/SRC_UniOvi/refs/heads/main/ML-IA/data/LoanDefault/X_Test_PCA_LoanDefault.pkl')
Y_train = pd.read_pickle('https://raw.githubusercontent.com/sranillac/SRC_UniOvi/refs/heads/main/ML-IA/data/LoanDefault/Y_Train_LoanDefault.pkl')
Y_test = pd.read_pickle('https://raw.githubusercontent.com/sranillac/SRC_UniOvi/refs/heads/main/ML-IA/data/LoanDefault/Y_Test_LoanDefault.pkl')
print(X_train.shape, X_test.shape)
print(Y_train.shape, Y_test.shape)
```

#### ``Evaluation Metrics``

1) ``Confusion matrix`` is a table that allows us to visualize the performance of a classification model:  
    - True Positives (**TP**): number of instances correctly predicted as positive
    - True Negatives (**TN**): number of instances correctly predicted as negative
    - False Positives (**FP**): number of instances incorrectly predicted as positive
    - False Negatives (**FN**): number of instances incorrectly predicted as negative.  


2) ``Accuracy`` measures the proportion of correct predictions (both positive and negative) out of the total predictions made:  

    $\text{Accuracy} = \frac{TP + TN}{TP + TN + FP + FN}$  

3) ``Precision`` measures the proportion of positive predictions that are actually correct:  

    $\text{Precision} = \frac{TP}{TP + FP}$  

4) ``Recall`` measures the proportion of actual positive instances that were correctly identified by the model:  

     $\text{Recall} = \frac{TP}{TP + FN}$  

5) ``F1-Score`` is the harmonic mean between **Precision** and **Recall**. It is useful when there is an imbalance in the classes:

     $\text{F1-Score} = 2 \times \frac{\text{Precision} \times \text{Recall}}{\text{Precision} + \text{Recall}}$

```python
# Evaluation Metrics function
def eval_metrics(y, y_pred):
    mc = confusion_matrix(y, y_pred)
    acc = accuracy_score(y, y_pred)
    cr = classification_report(y, y_pred)
    return mc, acc, cr
```

```python
# Logistic Regression Model - Evaluation
mc_LR, acc_LR, cr_LR = eval_metrics(Y_test, Y_pred_LR)

print(f'Logistic Regression Model - Confusion Matrix:')
print(mc_LR)
print('')
print(f'Logistic Regression Model - Accuracy: {acc_LR:.2f}')
print('')
print(f'Logistic Regression Model - Classification Report:')
print(cr_LR)
```

```python
# SVM Model - Evaluation
mc_SVM, acc_SVM, cr_SVM = eval_metrics(Y_test, Y_pred_SVM)

print(f'SVM Model - Confusion Matrix:')
print(mc_SVM)
print('')
print(f'SVM Model - Accuracy: {acc_SVM:.2f}')
print('')
print(f'SVM Model - Classification Report:')
print(cr_SVM)
```

```python
# K-Neighbors Model - Evaluation
mc_KNN, acc_KNN, cr_KNN = eval_metrics(Y_test, Y_pred_KNN)

print(f'KNN Model - Confusion Matrix:')
print(mc_KNN)
print('')
print(f'KNN Model - Accuracy: {acc_KNN:.2f}')
print('')
print(f'KNN Model - Classification Report:')
print(cr_KNN)
```

```python
# Decision Tree Model - Evaluation
mc_DT, acc_DT, cr_DT = eval_metrics(Y_test, Y_pred_DT)

print(f'Decision Tree Model - Confusion Matrix:')
print(mc_DT)
print('')
print(f'Decision Tree Model - Accuracy: {acc_DT:.2f}')
print('')
print(f'Decision Tree Model - Classification Report:')
print(cr_DT)
```

```python
# Random Forest Model - Evaluation
mc_RF, acc_RF, cr_RF = eval_metrics(Y_test, Y_pred_RF)

print(f'Random Forest Model - Confusion Matrix:')
print(mc_RF)
print('')
print(f'Random Forest Model - Accuracy: {acc_RF:.2f}')
print('')
print(f'Random Forest Model - Classification Report:')
print(cr_RF)
```

Which is the **``best performing model``**?  

- The **Logistic Regression** model achieves $89\%$ accuracy, with a strong precision of $0.89$ for **class 0**. However, its recall for **class 1** is very low ($0.02$), meaning it struggles to correctly identify positive cases.  

- The **SVM**, **Decision Tree** and **Random Forest** models all have an accuracy $88\%$ accuracy, but they completely fail to classify **class 1**, with a recall of $0$.  

- The **KNN** model also has an accuracy of 0.88 ($88\%$), but it performs slightly better than the other models in recognizing **class 1**, with a recall of $0.04$, though still very low.  

Therefore, **``Logistic Regression``** is the best model despite its limitations, offering the highest accuracy, but a very poor recall for **class 1**. It requires significant improvement—possibly through techniques like class balancing, feature engineering, or model tuning.
